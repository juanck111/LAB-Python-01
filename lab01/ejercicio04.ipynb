```python
import math

# --- Tuplas ---
rectangulo = (10, 5) # ancho, alto
area = rectangulo[0] * rectangulo[1]
perimetro = 2 * (rectangulo[0] + rectangulo[1])
# rectangulo[0] = 15 # Esto generaría un TypeError: 'tuple' object does not support item assignment. Son inmutables.

posiciones = [(0, 0), (1, 2), (3, 4)]
distancia_total = sum(math.dist(posiciones[i], posiciones[i+1]) for i in range(len(posiciones)-1))
print(f"Distancia total: {distancia_total:.2f}")

# --- Biblioteca ---
biblioteca = {
    "libros": [
        {"titulo": "1984", "autor": "George Orwell", "prestado": False},
        {"titulo": "El Principito", "autor": "Antoine de Saint-Exupéry", "prestado": True},
    ],
    "usuarios": ["Ana", "Luis", "María"]
}

prestados = sum(1 for libro in biblioteca["libros"] if libro["prestado"])
disponibles = len(biblioteca["libros"]) - prestados
print(f"Prestados: {prestados}, Disponibles: {disponibles}")

biblioteca["libros"].append({"titulo": "Cien años de soledad", "autor": "Gabriel G. Márquez", "prestado": False})

for libro in biblioteca["libros"]:
    if libro["titulo"] == "1984":
        libro["prestado"] = True

print(biblioteca)
```
