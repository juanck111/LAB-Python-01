```python
inventario = [
    {"nombre": "Pan", "precio": 1500, "stock": 30},
    {"nombre": "Leche", "precio": 3500, "stock": 15},
    {"nombre": "Café", "precio": 8000, "stock": 10},
]

print(f"Primer producto: {inventario[0]['nombre']}")

# Agrega nuevo producto
inventario.append({"nombre": "Huevos", "precio": 600, "stock": 60})

# Disminuye stock de Leche
for p in inventario:
    if p["nombre"] == "Leche":
        p["stock"] -= 2

# Valor total
valor_total = sum(p["precio"] * p["stock"] for p in inventario)
print(f"Valor total del inventario: ${valor_total}")

# Menor a 10 en stock
bajo_stock = [p for p in inventario if p["stock"] < 10]
print(f"Productos con bajo stock: {bajo_stock}")

# Aumento de precio 10% y redondeo
for p in inventario:
    p["precio"] = round(p["precio"] * 1.10, 2)

# Búsqueda
buscar = input("Ingrese el producto a buscar: ").capitalize()
encontrado = next((p for p in inventario if p["nombre"] == buscar), "No encontrado")
print(f"Resultado búsqueda: {encontrado}")

# Elimina stock 0 y ordena
inventario = [p for p in inventario if p["stock"] > 0]
inventario.sort(key=lambda x: x["precio"])
```
