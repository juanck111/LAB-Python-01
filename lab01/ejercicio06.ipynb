```python
n = int(input("Ingrese la cantidad de números naturales a sumar: "))
suma_n = sum(range(1, n + 1))
print(f"Suma de los primeros {n} números: {suma_n}")

acumulador = 0
for i in range(1, 6):
    acumulador += i
    if i % 2 == 0:
        print(f"Par encontrado: {i}")
print(f"Acumulador final: {acumulador}")

# Temperaturas mayores al promedio del Ejercicio 1
temperaturas = [23.7, 23.0, 21.8, 24.3, 25.1, 26.0] # Estado tras Ej1
promedio = sum(temperaturas) / len(temperaturas)
mayores_prom = [t for t in temperaturas if t > promedio]
print(f"Temperaturas mayores al promedio ({promedio:.2f}): {mayores_prom}")
```
