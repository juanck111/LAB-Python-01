```python
import requests

nombre_pokemon = input("Ingrese el nombre de un Pokémon: ").lower()
url = f"https://pokeapi.co/api/v2/pokemon/{nombre_pokemon}"
respuesta = requests.get(url)

if respuesta.status_code == 200:
    datos = respuesta.json()
    pokemon = {
        "nombre": datos["name"],
        "altura": datos["height"],
        "peso": datos["weight"],
        "tipos": [t["type"]["name"] for t in datos["types"]],
        "habilidades": [h["ability"]["name"] for h in datos["abilities"]],
        "experiencia_base": datos["base_experience"] # Otro uso interesante
    }

    print("\n📘 Información del Pokémon")
    print(f"Nombre: {pokemon['nombre'].capitalize()}")

    # Condicionales adicionales
    if len(pokemon["tipos"]) > 1:
        print(f"Tipo: Mixto ({', '.join(pokemon['tipos'])})")
    else:
        print(f"Tipo: Único ({pokemon['tipos'][0]})")

    if pokemon["peso"] > 100:
        print(f"Contextura: Pokémon de gran tamaño (Peso: {pokemon['peso']})")
    else:
        print(f"Contextura: Pokémon ligero (Peso: {pokemon['peso']})")

    print("Habilidades:")
    for habilidad in pokemon["habilidades"]:
        print(f"  - {habilidad.capitalize()}")

    print(f"Experiencia base: {pokemon['experiencia_base']}")

else:
    print("Pokémon no encontrado. Verifique el nombre ingresado.")
```
