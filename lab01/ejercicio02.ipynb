```python
temperaturas = [22.5, 23.0, 21.8, 24.3, 25.1]

temperaturas.append(26.0) # Agrega al final
temperaturas.pop(0)       # Elimina la primera posición
temperaturas.insert(1, 23.7) # Inserta en la segunda posición (índice 1)

temperaturas_ordenadas = sorted(temperaturas)
print(f"Lista ordenada: {temperaturas_ordenadas}")

diferencia = max(temperaturas) - min(temperaturas)
promedio = sum(temperaturas) / len(temperaturas)
print(f"Diferencia max-min: {diferencia:.2f}, Promedio: {promedio:.2f}")

indice_buscado = temperaturas.index(23.7)
print(f"Índice de 23.7: {indice_buscado}")

print(f"De la 2da a la 4ta posición: {temperaturas[1:4]}")
print(f"Tres primeras temperaturas: {temperaturas[:3]}")
```
